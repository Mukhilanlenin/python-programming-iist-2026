# Task 1: Physical Fitness & Medical Eligibility Checker

### Title & Objective
Write a program that reads a candidate's details, checks them against the standards, and prints either **ELIGIBLE** or **NOT ELIGIBLE with every reason for rejection**.

### Prerequisites / Key Constructs
`input()`, `int()`, `float()`, `if`, `elif`/`else`, comparison operators (`>=`, `<=`, `!=`), **`and` / `or` / `not`**, a list with `.append()`, `for` loop, `len()`, f-strings.

### Problem Statement / Context
A recruitment desk must screen candidates quickly. Illustrative standards:

| Parameter | MALE | FEMALE |
|---|---|---|
| Age | 18 - 25 years | 18 - 30 years |
| Minimum height | 170 cm | 157 cm |
| Education | at least Class 10 | at least Class 8 |
| Medical | Must be fit | Must be fit |

The program must report **all** failed conditions, not just the first one.

### Source Code Structure (Sample Solution)

````python
print("=" * 45)
print("     RECRUITMENT ELIGIBILITY CHECKER")
print("=" * 45)

# Step 1: Input
name      = input("Candidate name         : ")
gender    = input("Gender (MALE/FEMALE)   : ").upper()
age       = int(input("Age (years)            : "))
height    = float(input("Height (cm)            : "))
education = int(input("Highest class passed   : "))
medical   = input("Medically fit (yes/no) : ").lower()

# Step 2: Decide the limits using if-else based on Gender
if gender == "FEMALE":
    max_age    = 30
    min_height = 157
    edu = 8

else:  # Defaults to MALE requirements
    max_age    = 25
    min_height = 170
    edu = 10

# Step 3: Check every condition, remember reasons
reasons = []

if age < 18 or age > max_age:
    reasons.append(f"Age {age} is outside 18-{max_age}")

if height < min_height:
    reasons.append(f"Height {height} cm is below {min_height} cm")

if education < edu:
    reasons.append("Education is below Class 10")

if medical != "yes":
    reasons.append("Medically unfit")

# Step 4: Report
print("-" * 45)
print("Candidate :", name)
if len(reasons) == 0:
    print("RESULT    : ELIGIBLE for the Physical Standard Test")
else:
    print("RESULT    : NOT ELIGIBLE")
    for r in reasons:
        print("   ->", r)
````

# Task 2: Assessment Grade Calculator

### Title & Objective
Read the marks of a trainee in five training subjects, then calculate **total, percentage, highest and lowest subject, and final grade**.

### Prerequisites / Key Constructs
`for` loop over a list, `while` loop for input validation, `if / elif / else` ladder, counters and accumulators (`total += marks`), `and` / `or`, f-string formatting `:.2f`.

### Problem Statement / Context
Each subject is out of 100. A trainee **fails** if any subject is below **33**, whatever the total. Otherwise the grade depends on percentage:

| Percentage | Grade |
|---|---|
| 75 and above | A (Distinction) |
| 60 to 74.99 | B (First Class) |
| 45 to 59.99 | C (Second Class) |
| Below 45 (all subjects passed) | D (Pass) |

Marks outside 0 - 100 must be rejected and asked again.

### Source Code Structure (Sample Solution)

````python
subjects = ["Weapon Handling", "Map Reading", "Law & Procedure",
            "Field Craft", "Physical Training"]

total        = 0
failed_count = 0
highest      = -1
lowest       = 101
best_subject  = ""
worst_subject = ""

name = input("Trainee name: ")
print("Enter marks out of 100")

for subject in subjects:
    marks = int(input(f"  {subject:<18}: "))

    # Validation: keep asking until marks are between 0 and 100
    while marks < 0 or marks > 100:
        print("  Invalid! Marks must be between 0 and 100.")
        marks = int(input(f"  {subject:<18}: "))

    total = total + marks

    if marks < 33:
        failed_count = failed_count + 1
    if marks > highest:
        highest = marks
        best_subject = subject
    if marks < lowest:
        lowest = marks
        worst_subject = subject

percentage = total / (len(subjects) * 100) * 100

# Grade ladder: the fail rule is checked FIRST
if failed_count > 0:
    grade = "FAIL (below 33 in a subject)"
elif percentage >= 75:
    grade = "A (Distinction)"
elif percentage >= 60:
    grade = "B (First Class)"
elif percentage >= 45:
    grade = "C (Second Class)"
else:
    grade = "D (Pass)"

print("\n" + "=" * 40)
print(f"Trainee    : {name}")
print(f"Total      : {total} / {len(subjects) * 100}")
print(f"Percentage : {percentage:.2f}%")
print(f"Strongest  : {best_subject} ({highest})")
print(f"Weakest    : {worst_subject} ({lowest})")
print(f"Subjects failed: {failed_count}")
print(f"GRADE      : {grade}")
print("=" * 40)
````

# Task 3: Unit & Distance Converter

### Title & Objective
Build a menu-driven converter used for patrol planning and logistics that keeps running until the user chooses **Exit**.

### Prerequisites / Key Constructs
`while True`, **`break`**, **`continue`**, `if / elif` ladder, `in` with a list, constants, `float()`, f-string with `:g` and `:.2f`, counter variable.

### Problem Statement / Context
A patrol commander receives maps in kilometres, a foreign liaison officer speaks in miles, weather reports use both Celsius and Fahrenheit, and cargo is weighed in kg or lb. The converter must:
- show a menu repeatedly,
- reject invalid menu choices without crashing (`continue`),
- exit cleanly with a summary (`break`).

Formulas: `miles = km x 0.621371`, `F = C x 9/5 + 32`, `lb = kg x 2.20462`

### Source Code Structure (Sample Solution)

````python
KM_TO_MILES = 0.621371
KG_TO_LB    = 2.20462

conversions = 0

while True:
    print("\n===== UNIT & DISTANCE CONVERTER =====")
    print("1. Kilometres -> Miles")
    print("2. Miles -> Kilometres")
    print("3. Celsius -> Fahrenheit")
    print("4. Fahrenheit -> Celsius")
    print("5. Kilograms -> Pounds")
    print("6. Pounds -> Kilograms")
    print("7. Exit")
    choice = input("Enter choice (1-7): ")

    if choice == "7":
        print(f"Conversions done: {conversions}.")
        break

    if choice not in ["1", "2", "3", "4", "5", "6"]:
        print("Invalid choice. Please enter 1 to 7.")
        continue

    value = float(input("Enter value to convert: "))

    if choice == "1":
        print(f"{value:g} km = {value * KM_TO_MILES:.2f} miles")
    elif choice == "2":
        print(f"{value:g} miles = {value / KM_TO_MILES:.2f} km")
    elif choice == "3":
        print(f"{value:g} C = {value * 9 / 5 + 32:.1f} F")
    elif choice == "4":
        print(f"{value:g} F = {(value - 32) * 5 / 9:.1f} C")
    elif choice == "5":
        print(f"{value:g} kg = {value * KG_TO_LB:.2f} lb")
    elif choice == "6":
        print(f"{value:g} lb = {value / KG_TO_LB:.2f} kg")

    conversions = conversions + 1
````

# Task 4: Personnel Information System

### Title & Objective
Capture the service details of several personnel, validate the entries, and print a **formatted profile card** for each along with a small summary.

### Prerequisites / Key Constructs
`for` + `range()`, `while` validation loop, **lists** with `.append()` and indexing, f-string alignment (`:<25`, `:^36`), string multiplication (`"-" * 38`), finding the maximum in a list.

### Problem Statement / Context
A unit clerk maintains a quick roll of personnel. For each person capture **name, rank, service number, unit, age and years of service**. Rules:
- Age must be between **18 and 60**.
- Years of service cannot be negative and cannot exceed `age - 18`.
- Retirement age is **60**, so show the years remaining.
- After all cards, report the **senior-most person** (maximum years of service).

### Source Code Structure (Sample Solution)

````python
RETIREMENT_AGE = 60

names, service_nos, ages, services = [], [], [], []

count = int(input("How many personnel to enter? "))

# Step 1: Data entry with validation
for i in range(1, count + 1):
    print(f"\n--- Personnel {i} ---")
    name       = input("Name            : ")
    service_no = input("Service number  : ")
   

    age = int(input("Age (18-60)     : "))
    while age < 18 or age > 60:
        print("Invalid age! Enter a value from 18 to 60.")
        age = int(input("Age (18-60)     : "))

    years = int(input("Years of service: "))
    while years < 0 or years > age - 18:
        print("Invalid! Service years must be between 0 and", age - 18)
        years = int(input("Years of service: "))

    names.append(name)
    service_nos.append(service_no)
    ages.append(age)
    services.append(years)

# Step 2: Print profile cards
print("\n" + "#" * 40)
for i in range(count):
    remaining = RETIREMENT_AGE - ages[i]
    print("+" + "-" * 38 + "+")
    print(f"| {'PERSONNEL PROFILE CARD':^36} |")
    print("+" + "-" * 38 + "+")
    print(f"| {'Name':<10}: {names[i]:<24} |")
    print(f"| {'Service No':<10}: {service_nos[i]:<24} |")
    print(f"| {'Age':<10}: {ages[i]:<24} |")
    print(f"| {'Service':<10}: {str(services[i]) + ' years':<24} |")
    print(f"| {'To retire':<10}: {str(remaining) + ' years':<24} |")
    print("+" + "-" * 38 + "+")

# Step 3: Senior-most person
senior = 0
for i in range(1, count):
    if services[i] > services[senior]:
        senior = i
print(f"\nSenior-most: {names[senior]} ({services[senior]} years)")
````

### Problem Statement / Context
A store officer records daily ration issue counts for a company. Write a program that:
1. reads **N** numbers into a list,
2. prints all **even** and **odd** numbers separately with counts,
3. finds **sum, average, maximum and minimum** using loops,
4. counts how many numbers are **above the average**,
5. prints the **sum of the first N natural numbers** using `while` and checks it with the formula `N(N+1)/2`,
6. prints a **multiplication table** of any number.

### Source Code Structure (Sample Solution)

````python
# Step 1: Read numbers into a list
n = int(input("How many numbers? "))
numbers = []
for i in range(n):
    value = int(input(f"Enter number {i + 1}: "))
    numbers.append(value)
print("\nList:", numbers)

````

````python
# Step 2: One pass to compute everything
total   = 0
maximum = numbers[0]
minimum = numbers[0]
evens, odds = [], []

for num in numbers:
    total = total + num  # total
    if num % 2 == 0:
        evens.append(num)
    else:
        odds.append(num)
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num

average = total / n     # average

above_avg = 0
for num in numbers:
    if num > average:
        above_avg = above_avg + 1

print("Even numbers :", evens, "-> count", len(evens))
print("Odd numbers  :", odds, "-> count", len(odds))
print("Sum          :", total)
print(f"Average      : {average:.2f}")
print("Maximum      :", maximum)
print("Minimum      :", minimum)
print("Above average:", above_avg, "numbers")
````

````python
# Step 3: Sum of first N natural numbers using while
limit = int(input("\nSum of first how many natural numbers? "))
i = 1
s = 0
while i <= limit:
    s = s + i
    i = i + 1
print(f"Sum of 1 to {limit} = {s}  (formula check: {limit * (limit + 1) // 2})")
````

````python
# Step 4: Multiplication table
t = int(input("\nPrint table of: "))
for k in range(1, 11):
    print(f"{t} x {k:>2} = {t * k}")
````

# Task 6: Master Menu-Driven Program

### Title & Objective
Combine **everything learned today** into one menu-driven number toolkit. The user enters a number once and then applies several operations to it until choosing to exit.

### Prerequisites / Key Constructs
`while True`, **`break`**, **`continue`**, `if / elif / else`, `for` with `range()`, nested `while` (digit extraction using `%` and `//`), a **flag variable** (`is_prime`), `and` / `or`, input validation.

### Problem Statement / Context
The intelligence unit's analysts often test identification numbers and codes. Build a toolkit with these options:

| Option | Operation | Idea |
|---|---|---|
| 1 | Even / Odd | `n % 2` |
| 2 | Factorial | multiply `1..n` in a loop |
| 3 | Sum of digits | `n % 10` and `n // 10` in a `while` loop |
| 4 | Reverse the number | `rev = rev * 10 + digit` |
| 5 | Prime check | try divisors up to the square root, use `break` |
| 6 | Multiplication table | `for` loop |
| 7 | Enter a new number | reassign `number` |
| 8 | Exit | `break` |

### Source Code Structure (Sample Solution)

````python
# Get a valid starting number
number = int(input("Enter a positive whole number: "))
while number <= 0:
    print("Number must be greater than zero.")
    number = int(input("Enter a positive whole number: "))

operations = 0

while True:
    print(f"\n======= NUMBER TOOLKIT (working on {number}) =======")
    print("1. Even / Odd")
    print("2. Factorial")
    print("3. Sum of digits")
    print("4. Reverse the number")
    print("5. Prime check")
    print("6. Multiplication table")
    print("7. Enter a new number")
    print("8. Exit")
    choice = input("Choose (1-8): ")

    if choice == "8":
        print(f"Operations performed: {operations}. Session closed.")
        break

    elif choice == "1": # Even/odd
        if number % 2 == 0:
            print(number, "is EVEN")
        else:
            print(number, "is ODD")

    elif choice == "2": #Factorial
        fact = 1
        for k in range(1, number + 1):
            fact = fact * k
        print(f"{number}! = {fact}")

    elif choice == "3": #Sum of digits
        temp = number
        digit_sum = 0
        while temp > 0:
            digit_sum = digit_sum + temp % 10
            temp = temp // 10
        print("Sum of digits =", digit_sum)

    elif choice == "4": # Reverse the number
        temp = number
        rev = 0
        while temp > 0:
            rev = rev * 10 + temp % 10
            temp = temp // 10
        print("Reversed number =", rev)

    elif choice == "5": # Prime check
        is_prime = True
        if number < 2:
            is_prime = False
        else:
            for d in range(2, int(number ** 0.5) + 1):
                if number % d == 0:
                    is_prime = False
                    break
        if is_prime:
            print(number, "is a PRIME number")
        else:
            print(number, "is NOT a prime number")

    elif choice == "6": # Multiplication table
        for k in range(1, 11):
            print(f"{number} x {k:>2} = {number * k}")

    elif choice == "7":
        number = int(input("Enter the new number: "))
        while number <= 0:
            print("Number must be greater than zero.")
            number = int(input("Enter the new number: "))

    else:
        print("Invalid choice! Please choose 1 to 8.")
        continue

    operations = operations + 1
````
