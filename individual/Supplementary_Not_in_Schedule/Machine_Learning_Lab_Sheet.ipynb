# 4. Linear Regression 

Topics: simple linear regression by gradient descent, and by least squares (statistics formulae and linear algebra).

Model: $y = b_0 + b_1 x$ (or $y = mx + b$). Cost: $J(b,m)=\frac1N\sum_i (y_i-(mx_i+b))^2$.
Gradient: $\partial J/\partial m = -\frac2N\sum x_i(y_i-(mx_i+b))$, $\partial J/\partial b = -\frac2N\sum (y_i-(mx_i+b))$.
Least squares: $b_1 = \mathrm{Cov}(x,y)/\mathrm{Var}(x)$, $b_0=\bar y-b_1\bar x$; normal equations: $b=(X^TX)^{-1}X^TY$.

## 4.1 Gradient descent from scratch

### Example

```python
import numpy as np, matplotlib.pyplot as plt
%matplotlib inline
try:
    points = np.genfromtxt('score.csv', delimiter=',')      # hours studied, test score
except OSError:                                              # fallback synthetic data
    rng = np.random.default_rng(0); xs = rng.uniform(20,80,100)
    points = np.c_[xs, 1.3*xs + 8 + rng.normal(0,10,100)]
x, y = points[:,0], points[:,1]

def compute_cost(b, m, points):
    return np.mean((points[:,1] - (m*points[:,0] + b))**2)

def step_gradient(b, m, points, lr):
    N = len(points); X, Y = points[:,0], points[:,1]
    err = Y - (m*X + b)
    return b - lr*(-2/N)*err.sum(), m - lr*(-2/N)*(X*err).sum()

def gradient_descent_runner(points, b, m, lr, n_iter):
    cost = []
    for _ in range(n_iter):
        cost.append(compute_cost(b, m, points))
        b, m = step_gradient(b, m, points, lr)
    return b, m, cost

b, m, cost = gradient_descent_runner(points, 0, 0, 0.0001, 1000)
print("b =", b, " m =", m, " final cost =", compute_cost(b, m, points))
plt.plot(cost); plt.xlabel('iterations'); plt.ylabel('cost'); plt.show()
plt.scatter(points[:,0], points[:,1]); plt.plot(points[:,0], m*points[:,0]+b, 'r'); plt.show()
```

## 4.2 Least squares: statistics formulae and linear algebra

### Example

```python
import pandas as pd
from sklearn.model_selection import train_test_split
rng = np.random.default_rng(1); xs = rng.integers(1,120,60)
data = pd.DataFrame({'x':xs, 'y':3.4*xs + 19 + rng.normal(0,25,60)})   # stand-in for insurance.csv
x = data['x'].values; y = data['y'].values
x_tr, x_te, y_tr, y_te = train_test_split(x, y, test_size=0.2, random_state=42)

b1 = np.cov(x_tr, y_tr)[0,1] / np.var(x_tr, ddof=1)
b0 = y_tr.mean() - b1*x_tr.mean()
rmse = np.sqrt(np.mean((y_te - (b0 + b1*x_te))**2))
print("b0, b1 =", b0, b1, " test RMSE =", rmse)

# Linear-algebra solution: b = (X^T X)^-1 X^T Y
X = np.column_stack([np.ones_like(x_tr), x_tr])
b = np.linalg.pinv(X.T @ X) @ (X.T @ y_tr)
print("normal equations:", b)
```

# 5. Principal Component Analysis 

PCA finds the directions of maximum variance and projects the data onto a lower-dimensional subspace.
Steps: (1) standardize; (2) eigendecomposition of the covariance/correlation matrix (or SVD); (3) sort eigenpairs and pick top $k$; (4) build projection matrix $W$; (5) $Y = XW$.
Covariance: $\Sigma=\frac{1}{n-1}(X-\bar x)^T(X-\bar x)$. Dataset: Iris (150 samples, 4 features, 3 classes).

### Example

```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

iris = load_iris(); X, y = iris.data, iris.target
X_std = StandardScaler().fit_transform(X)

cov = np.cov(X_std.T)
eig_vals, eig_vecs = np.linalg.eigh(cov)
idx = np.argsort(eig_vals)[::-1]; eig_vals, eig_vecs = eig_vals[idx], eig_vecs[:,idx]
var_exp = 100*eig_vals/eig_vals.sum()
print("eigenvalues:", eig_vals); print("explained variance %:", var_exp, np.cumsum(var_exp))

W = np.hstack([eig_vecs[:,0].reshape(-1,1), eig_vecs[:,1].reshape(-1,1)])   # 4x2 projection matrix
Y = X_std @ W
for lab, col in zip(range(3), ('blue','red','green')):
    plt.scatter(Y[y==lab,0], Y[y==lab,1], c=col, label=iris.target_names[lab])
plt.xlabel('PC1'); plt.ylabel('PC2'); plt.legend(); plt.show()

Y_sk = PCA(n_components=2).fit_transform(X_std)      # shortcut in scikit-learn
```

# 6. Support Vector Machines 

SVMs are discriminative classifiers: they choose the separating line that **maximizes the margin**. Only the support vectors affect the fit. Kernels (e.g. RBF) allow non-linear boundaries; the parameter $C$ controls how hard the margin is (large $C$ = hard, small $C$ = soft).

### Example

```python
import numpy as np, matplotlib.pyplot as plt, seaborn as sns; sns.set()
from sklearn.datasets import make_blobs, make_circles
from sklearn.svm import SVC

def plot_svc_decision_function(model, ax=None, plot_support=True):
    ax = ax or plt.gca()
    xlim, ylim = ax.get_xlim(), ax.get_ylim()
    xx = np.linspace(*xlim, 30); yy = np.linspace(*ylim, 30)
    YY, XX = np.meshgrid(yy, xx)
    P = model.decision_function(np.vstack([XX.ravel(), YY.ravel()]).T).reshape(XX.shape)
    ax.contour(XX, YY, P, colors='k', levels=[-1,0,1], alpha=0.5, linestyles=['--','-','--'])
    if plot_support:
        ax.scatter(model.support_vectors_[:,0], model.support_vectors_[:,1], s=100, lw=1, facecolors='none')
    ax.set_xlim(xlim); ax.set_ylim(ylim)

# Maximum-margin linear SVM
X, y = make_blobs(n_samples=50, centers=2, random_state=0, cluster_std=0.60)
model = SVC(kernel='linear', C=1E10).fit(X, y)
plt.scatter(X[:,0], X[:,1], c=y, s=50, cmap='autumn'); plot_svc_decision_function(model); plt.show()
print(model.support_vectors_)

# Kernel SVM on data that is not linearly separable
X, y = make_circles(100, factor=.1, noise=.1)
clf = SVC(kernel='rbf', C=1E6).fit(X, y)
plt.scatter(X[:,0], X[:,1], c=y, s=50, cmap='autumn'); plot_svc_decision_function(clf); plt.show()
```

# 7. Bias–Variance Tradeoff

Learning objectives: understand bias, variance and irreducible error; visualize the tradeoff; implement the decomposition; see how complexity affects underfitting/overfitting.

$$\text{Expected Error}=\text{Bias}^2+\text{Variance}+\text{Irreducible Error}$$

- **Bias**: error from wrong assumptions (underfitting) — $\text{Bias}^2=E[(\hat f(x)-f(x))^2]$ averaged over training sets.
- **Variance**: sensitivity to the training sample (overfitting) — $E[(\hat f(x)-E[\hat f(x)])^2]$.
- **Irreducible error**: noise $\sigma^2$.

Higher complexity ⇒ bias↓, variance↑. Goal: the sweet spot of minimum total error. True function used: $f(x)=x\sin x$, $y=f(x)+\varepsilon$.

### Example

```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
np.random.seed(42)

def generate_data(n_samples=100, noise_std=1.0):
    x = np.random.uniform(0, 10, n_samples)
    return x.reshape(-1,1), x*np.sin(x) + np.random.normal(0, noise_std, n_samples)

X_test = np.linspace(0, 10, 200).reshape(-1,1)
y_true = (X_test*np.sin(X_test)).ravel()          # keep 1-D to avoid broadcasting errors

def compute_bias_variance(make_model, X_test, y_true, n_train=50, noise_std=1.5, n_iter=200):
    # draw a FRESH training set each time from the data-generating process
    preds = np.zeros((n_iter, len(X_test)))
    for i in range(n_iter):
        Xtr, ytr = generate_data(n_train, noise_std)
        preds[i] = make_model().fit(Xtr, ytr).predict(X_test)
    bias2 = np.mean((preds.mean(axis=0) - y_true)**2)
    var = np.mean(preds.var(axis=0))
    return bias2, var, bias2 + var

degrees = [1, 2, 3, 5, 8, 12]
res = [compute_bias_variance(lambda d=d: make_pipeline(PolynomialFeatures(d), LinearRegression()), X_test, y_true)
       for d in degrees]
b2, v, tot = zip(*res)
plt.plot(degrees, b2, 'o-', label='Bias$^2$'); plt.plot(degrees, v, 'o-', label='Variance')
plt.plot(degrees, tot, 'o-', label='Total'); plt.xlabel('Polynomial degree'); plt.legend(); plt.grid(alpha=.3); plt.show()
```

# 8. Kernel Perceptron

Learning objectives: dual representation of the perceptron; derive the kernel perceptron; implement it from scratch; compare kernels; visualize non-linear decision boundaries.

**Derivation**
1. Classic perceptron update: if $y_i(w\cdot x_i)\le 0$ then $w\leftarrow w+y_ix_i$.
2. Dual representation: after updates $w=\sum_i \alpha_i y_i x_i$, where $\alpha_i$ = number of times example $i$ caused an update.
3. Prediction: $\operatorname{sign}\left(\sum_i \alpha_i y_i\,(x_i\cdot x)\right)$.
4. Kernel trick: replace $x_i\cdot x$ with $K(x_i,x)$ ⇒ $\operatorname{sign}\left(\sum_i \alpha_i y_i K(x_i,x)\right)$.
5. Update: initialise $\alpha=0$; for each misclassified $i$: $\alpha_i\leftarrow\alpha_i+1$.

**Kernels:** linear $K=x_1\cdot x_2$; polynomial $K=(x_1\cdot x_2+c_0)^d$; RBF $K=\exp(-\gamma\|x_1-x_2\|^2)$.

### Example

```python
import numpy as np, matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from sklearn.datasets import make_circles, make_moons
np.random.seed(42)

def plot_decision_boundary(model, X, y, title="Decision Boundary", resolution=0.05):
    x_min, x_max = X[:,0].min()-0.5, X[:,0].max()+0.5
    y_min, y_max = X[:,1].min()-0.5, X[:,1].max()+0.5
    xx, yy = np.meshgrid(np.arange(x_min, x_max, resolution), np.arange(y_min, y_max, resolution))
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    plt.figure(figsize=(8,5))
    plt.contourf(xx, yy, Z, alpha=0.3, cmap=ListedColormap(['#FFAAAA','#AAAAFF']))
    plt.scatter(X[:,0], X[:,1], c=y, cmap=ListedColormap(['red','blue']), edgecolors='k', s=60)
    if getattr(model, 'alpha', None) is not None:
        sv = model.alpha > 0
        plt.scatter(X[sv,0], X[sv,1], s=150, facecolors='none', edgecolors='green', linewidths=2,
                    label=f'Support vectors ({sv.sum()})'); plt.legend()
    plt.title(title); plt.tight_layout(); plt.show()

# Data sets used in the experiments
X_linear = np.random.randn(100, 2);  y_linear = np.where(X_linear.sum(1) > 0, 1, -1)
X_circ, y_circ = make_circles(n_samples=200, noise=0.1, factor=0.5, random_state=42); y_circ = np.where(y_circ==0, -1, 1)
X_moon, y_moon = make_moons(n_samples=200, noise=0.15, random_state=42);              y_moon = np.where(y_moon==0, -1, 1)
```

# 9. Naive Bayes Classifier

A Naive Bayes classifier applies Bayes' theorem with the 'naive' assumption that features are conditionally independent:

$$P(A\mid B)=\frac{P(B\mid A)\,P(A)}{P(B)}$$

Performance is judged by accuracy, precision, recall and F-score. The exercises use the Pima Indians diabetes data (`diabetes.csv`, 768 rows, 8 features, `Outcome` = 0/1).

### Example

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

X, y = load_iris(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
gnb = GaussianNB().fit(X_tr, y_tr)
pred = gnb.predict(X_te)
print("accuracy:", accuracy_score(y_te, pred))
print(confusion_matrix(y_te, pred))
print(classification_report(y_te, pred))
```

# 10. Bayesian Inference: Dirichlet–Multinomial Conjugate Model

Learning objectives: Bayesian inference with conjugate priors; Dirichlet–Multinomial models in PyMC; effect of the prior on the posterior; visualizing posteriors; posterior predictive checks.

**Conjugacy.** Prior $\theta\sim\mathrm{Dir}(\alpha)$, counts $c\sim\mathrm{Multinomial}(n,\theta)$ ⇒ posterior $\theta\mid c\sim\mathrm{Dir}(\alpha+c)$.
- Posterior mean: $E[\theta_i\mid c]=\dfrac{\alpha_i+c_i}{\sum_j(\alpha_j+c_j)}$
- MAP: $\dfrac{\alpha_i+c_i-1}{\sum_j(\alpha_j+c_j-1)}$ (for $\alpha_i+c_i>1$)

### Example

```python
import numpy as np
rng = np.random.default_rng(42)

# Dirichlet posterior update for a die rolled 10 times:  counts [4,3,3] with a uniform prior
alpha_prior = np.array([1., 1., 1.]); counts = np.array([4, 3, 3])
alpha_post = alpha_prior + counts
print("posterior mean:", alpha_post/alpha_post.sum())
print("MAP           :", (alpha_post-1)/(alpha_post-1).sum())
samples = rng.dirichlet(alpha_post, size=5000)           # analytical posterior samples
print("sample mean   :", samples.mean(axis=0))
# CAUTION: never reuse the name of your prior variable (e.g. as a loop variable) before sampling.
```

# 11. Density Estimation: Gaussian Mixture Models 

A Gaussian Mixture Model describes data as coming from $K$ Gaussian components. The **Expectation–Maximization (EM)** algorithm is an approach for maximum-likelihood estimation in the presence of latent variables:
- **E-step:** compute the responsibility (posterior probability) of each component for every point.
- **M-step:** re-estimate means, variances and mixing weights from those responsibilities.
- Repeat until the log-likelihood converges.

### Example

```python
import numpy as np, matplotlib.pyplot as plt, seaborn as sns
from sklearn.mixture import GaussianMixture
np.random.seed(66)

# Two overlapping 1-D processes (different from the exercise values)
d1 = np.random.normal(0, 1.5, 3000); d2 = np.random.normal(6, 1.0, 2000)
X = np.concatenate([d1, d2]).reshape(-1,1)
sns.histplot(X.ravel(), stat='density', bins=60, kde=True); plt.show()

gmm = GaussianMixture(n_components=2, n_init=5, random_state=66).fit(X)   # default k-means init, several restarts
print("means:", gmm.means_.ravel(), "variances:", gmm.covariances_.ravel(), "weights:", gmm.weights_)
print("converged:", gmm.converged_)
print("posterior for x=3:", gmm.predict_proba([[3.0]]))
print("cluster sizes:", np.bincount(gmm.predict(X)))
```

# 12. Maximum Likelihood Estimation 

MLE seeks the parameters $\theta$ that maximize the likelihood of the observed data:
$$L(X;\theta)=\prod_{i=1}^n P(x_i;\theta)\qquad\Rightarrow\qquad \text{maximize}\ \sum_{i=1}^n\log P(x_i;\theta)$$
(sums of log-probabilities are numerically stable).

For an exponential model $s_i\sim\mathrm{Exp}(\lambda)$: $L(\lambda)=\lambda^n\exp(-\lambda n\bar s)$, maximised at $\hat\lambda=1/\bar s$. Goodness of fit is assessed with the Kolmogorov–Smirnov test (distance between the empirical CDF and the model CDF).

### Example

```python
import numpy as np, scipy.stats as st, matplotlib.pyplot as plt
np.random.seed(0)
s = np.random.exponential(scale=200.0, size=300)             # synthetic survival-like data

rate = 1/s.mean()                                           # analytic MLE
print("rate (MLE):", rate, "  true rate:", 1/200)
loc, scale = st.expon.fit(s, floc=0)                        # numerical MLE with the location fixed at 0
print("scipy fit scale:", scale)

days = np.linspace(0, s.max(), 1000)
nbins = 30
plt.hist(s, nbins)
plt.plot(days, st.expon.pdf(days, scale=1/rate)*len(s)*s.max()/nbins, '-r', lw=3)
plt.xlabel('value'); plt.ylabel('count'); plt.show()

print(st.kstest(s, st.expon.cdf, (0, scale)))              # KS goodness-of-fit test
```

# 13. Decision Trees for Classification 

A decision tree recursively splits the feature space using a criterion (Gini impurity by default in scikit-learn; entropy/information gain via `criterion='entropy'`). scikit-learn implements an optimized version of **CART** (binary splits). Overfitting is controlled with stopping criteria such as `max_depth`, `min_samples_split`, `min_samples_leaf`, `max_leaf_nodes` or pruning (`ccp_alpha`).

### Example

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn import tree
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

iris = load_iris(as_frame=True)
df = iris.frame
X = df.drop('target', axis=1); y = df['target']
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=42)

clf = tree.DecisionTreeClassifier(max_depth=3, random_state=42).fit(X_tr, y_tr)
print("test accuracy:", clf.score(X_te, y_te))
plt.figure(figsize=(10,8))
tree.plot_tree(clf, feature_names=iris.feature_names, class_names=iris.target_names, filled=True)
plt.show()

new_flower = pd.DataFrame([[5, 1.5, 4, 2]], columns=iris.feature_names)
print(iris.target_names[clf.predict(new_flower)][0], clf.predict_proba(new_flower)[0])
```
