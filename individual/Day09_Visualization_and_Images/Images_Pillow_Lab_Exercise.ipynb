```python
from PIL import Image, ImageDraw
import numpy as np
from IPython.display import display
```

```python
img = Image.new("RGB", (320, 240), (135, 206, 235))
print(img.size, img.mode)
print("sky :", img.getpixel((10, 10)))
display(img)
```

```python
d = ImageDraw.Draw(img)
d.rectangle((100, 110, 220, 200), fill=(200, 120, 60))
d.polygon([(90, 110), (160, 60), (230, 110)],
fill=(170, 40, 40))
print("wall :", img.getpixel((160, 150)))
print("roof :", img.getpixel((160, 90)))
print("sky :", img.getpixel((10, 10)))
display(img)
```

```python
d.ellipse((240, 20, 290, 70), fill=(255, 210, 50))
print("sun :", img.getpixel((265, 45)))
img.save("scene.png")
img = Image.open("scene.png")
print(img.format, img.size, img.mode)
display(img)
```

```python
gray = img.convert("L") # one number per pixel
house = img.crop((90, 60, 230, 200)) # cut out the house
small = img.copy(); small.thumbnail((100, 100))
sq = img.resize((224, 224))
turn = img.rotate(90, expand=True)
display(gray, house, small, sq, turn)
```
